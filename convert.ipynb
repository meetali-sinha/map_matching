```python
import pandas as pd
import json

# Load the Excel file (adjust path and sheet name if needed)
df = pd.read_csv("updated_toll_data.csv")

# Create a list to store features
features = []

# Iterate through the rows of the DataFrame
for _, row in df.iterrows():
    feature = {
        "type": "Feature",
        "properties": {
            "name": row["name"],
            "LMV": row["LMV"],
            "MAV": row["MAV"]
        },
        "geometry": {
            "type": "Point",
            "coordinates": [row["lon"], row["lat"]]
        }
    }
    features.append(feature)

# Create the final GeoJSON structure
geojson = {
    "type": "FeatureCollection",
    "features": features
}

with open("toll_data.geojson", "w", encoding="utf-8") as f:
    json.dump(geojson, f, indent=4, ensure_ascii=False)


print("GeoJSON file created successfully.")
```

```python
import pandas as pd
import json

# Fix rupee symbol encoding
def fix_rupee_symbol(text):
    if isinstance(text, str):
        return text.replace('?', '₹')
    return text

# Read Excel file
df = pd.read_csv('updated_toll_data.csv')

# Convert to simple JSON format
toll_data = []

for _, row in df.iterrows():
    toll_entry = {
        "name": row['name'],
        "lat": row['lat'],
        "lon": row['lon'],
        "LMV": fix_rupee_symbol(row['LMV']),
        "MAV": fix_rupee_symbol(row['MAV'])
    }
    toll_data.append(toll_entry)

with open('toll_gates.json', 'w', encoding='utf-8') as f:
    json.dump(toll_data, f, indent=2, ensure_ascii=False)

print("Conversion completed!")
```

```python
import pandas as pd
import json

# Fix rupee symbol encoding
def fix_rupee_symbol(text):
    if isinstance(text, str):
        return text.replace('?', '₹')
    return text

# Read Excel file
df = pd.read_csv('updated_toll_data.csv')

# Convert to GeoJSON format
geojson = {
    "type": "FeatureCollection",
    "features": []
}

for _, row in df.iterrows():
    feature = {
        "type": "Feature",
        "geometry": {
            "type": "Point",
            "coordinates": [row['lon'], row['lat']]
        },
        "properties": {
            "name": row['name'],
            "LMV": fix_rupee_symbol(row['LMV']),
            "MAV": fix_rupee_symbol(row['MAV'])
        }
    }
    geojson["features"].append(feature)

# Save to JSON file with proper encoding
with open('toll_data.geojson', 'w', encoding='utf-8') as f:
    json.dump(geojson, f, indent=2, ensure_ascii=False)

print("Conversion completed!")
```
